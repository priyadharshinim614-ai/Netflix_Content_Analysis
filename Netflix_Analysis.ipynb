# Netflix Content Analysis


## Objective
Analyze Netflix movies and TV shows using python

```python
import pandas as pd
netflix_data=pd.read_csv("netflix_titles.csv")
netflix_data.head()
```

# Missing values Analysis 

```python
netflix_data.isnull().sum()
```

## Movies vs TV Shows on Netflix

```python
netflix_data['type'].value_counts()
```

```python
import matplotlib.pyplot as plt
netflix_data['type'].value_counts().plot(kind='bar')
plt.title("Movies vs TV shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

## Top 10 countries Analysis

```python
netflix_data['country'].value_counts().head(10)
```

```python
netflix_data['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Countries on Netflix")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

## Release Year trend Analysis

```python
netflix_data['release_year'].value_counts().head(10)
```

```python
netflix_data['release_year'].value_counts().sort_index().plot(figsize=(12,5))
plt.title("netflix content release trend")
plt.xlabel("release year")
plt.ylabel("number of titles")
plt.show()
```

## Netflix rating 

```python
netflix_data['rating'].value_counts().head(10)
```

```python
netflix_data['rating'].value_counts().head(10).plot(kind='bar')
plt.title("Netflix rating")
plt.xlabel("rating")
plt.ylabel("count")
plt.show()
```

## Conclusion
- Netflix contains more movies than TV shows.
- The dataset contains 8807 titles.
- Movies are the dominant content type on netflix.
